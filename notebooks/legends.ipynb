```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from pathlib import Path

# Output path
out_dir = Path("/storage/homefs/ge24z347/Zell_event/AGNO_PLOTS/Combiprecip_Agno_2m_precipitation")
out_dir.mkdir(parents=True, exist_ok=True)

out_file = out_dir / "legend_process_areas_agno.png"

fig, ax = plt.subplots(figsize=(5, 1.8))
ax.axis("off")

legend_elements = [
    Patch(facecolor="green", edgecolor="green", label="Surface runoff/groundwater flooding"),
    Patch(facecolor="magenta", edgecolor="magenta", label="flooding/debris flow"),
]

ax.legend(
    handles=legend_elements,
    loc="center",
    frameon=False
)

plt.savefig(out_file, dpi=300, bbox_inches="tight", transparent=True)
plt.show()

print("Saved to:", out_file)
```
