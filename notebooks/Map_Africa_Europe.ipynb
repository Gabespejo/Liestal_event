```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.io.shapereader as shpreader

# -----------------------------
# Create figure
# -----------------------------
fig = plt.figure(figsize=(12, 8))
ax = plt.axes(projection=ccrs.PlateCarree())

# Transparent background
fig.patch.set_alpha(0)
ax.set_facecolor('#e6e6e6')

# Extent
ax.set_extent([-60, 170, -50, 85], crs=ccrs.PlateCarree())

# -----------------------------
# Load Natural Earth countries
# -----------------------------
shapename = shpreader.natural_earth(
    resolution='110m',
    category='cultural',
    name='admin_0_countries'
)

reader = shpreader.Reader(shapename)

# -----------------------------
# Plot ALL countries (this is key!)
# -----------------------------
for country in reader.records():
    name = country.attributes['NAME_LONG']
    geom = country.geometry

    if name in ['Switzerland', 'Mozambique']:
        # Highlight
        ax.add_geometries(
            [geom],
            ccrs.PlateCarree(),
            facecolor='#d73027',
            edgecolor='black',
            linewidth=1.2,
            zorder=2
        )
    else:
        # Everything else in gray (including Americas!)
        ax.add_geometries(
            [geom],
            ccrs.PlateCarree(),
            facecolor='lightgray',
            edgecolor='none',
            alpha=0.7,
            zorder=1
        )

# -----------------------------
# Coastlines
# -----------------------------
ax.coastlines(linewidth=1, color='black')

# -----------------------------
# Labels
# -----------------------------
ax.text(8.5, 47.5, 'Switzerland', fontsize=10, weight='bold')
ax.text(36, -17, 'Mozambique', fontsize=10, weight='bold')

# -----------------------------
# Clean layout
# -----------------------------
ax.set_axis_off()
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.io.shapereader as shpreader

fig = plt.figure(figsize=(12, 8))
ax = plt.axes(projection=ccrs.PlateCarree())

# Background
fig.patch.set_alpha(0)
ax.set_facecolor('#f2f2f2')

ax.set_extent([-60, 170, -50, 85], crs=ccrs.PlateCarree())

shapename = shpreader.natural_earth(
    resolution='110m',
    category='cultural',
    name='admin_0_countries'
)

reader = shpreader.Reader(shapename)

for country in reader.records():
    name = country.attributes['NAME_LONG']
    geom = country.geometry

    if name in ['Switzerland', 'Mozambique']:
        ax.add_geometries(
            [geom],
            ccrs.PlateCarree(),
            facecolor='#e31a1c',
            edgecolor='black',
            linewidth=1.5,
            zorder=2
        )
    else:
        ax.add_geometries(
            [geom],
            ccrs.PlateCarree(),
            facecolor='#d9d9d9',
            edgecolor='none',
            alpha=1,
            zorder=1
        )

# Softer coastlines
ax.coastlines(linewidth=0.5, color='#666666')

# Labels (improved)
ax.text(9, 48, 'Switzerland', fontsize=12, weight='bold')
ax.text(37, -16, 'Mozambique', fontsize=12, weight='bold')

ax.set_axis_off()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.io.shapereader as shpreader

fig = plt.figure(figsize=(12, 8))
fig.patch.set_facecolor('white')   # map stays white

ax = plt.axes(projection=ccrs.PlateCarree())
ax.set_facecolor('white')

ax.set_extent([-60, 170, -50, 85], crs=ccrs.PlateCarree())

shapename = shpreader.natural_earth(
    resolution='110m',
    category='cultural',
    name='admin_0_countries'
)

reader = shpreader.Reader(shapename)

for country in reader.records():
    name = country.attributes['NAME_LONG']
    geom = country.geometry

    if name in ['Switzerland', 'Mozambique']:
        ax.add_geometries(
            [geom],
            ccrs.PlateCarree(),
            facecolor='#e31a1c',   # strong red
            edgecolor='black',
            linewidth=1.5,
            zorder=3
        )
    else:
        ax.add_geometries(
            [geom],
            ccrs.PlateCarree(),
            facecolor='#cfcfcf',   # slightly darker gray (better contrast)
            edgecolor='none',
            zorder=1
        )

ax.coastlines(linewidth=0.5, color='#666666')

ax.text(9, 48, 'Switzerland', fontsize=13, weight='bold')
ax.text(37, -16, 'Mozambique', fontsize=13, weight='bold')

ax.set_axis_off()
plt.tight_layout()
plt.show()
```
